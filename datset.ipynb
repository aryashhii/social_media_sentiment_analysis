```python
import pandas as pd


amazon_df = pd.read_csv("project.csv")
disaster_df = pd.read_csv("social.csv")
judge_df = pd.read_csv("datas.csv")
senti_df = pd.read_csv("final_sentiment_dataset.csv")


amazon_df = amazon_df[['clean_text', 'sentiment']]
disaster_df = disaster_df[['clean_text', 'sentiment']]
judge_df = judge_df[['clean_text', 'sentiment']]
senti_df = senti_df[['clean_text', 'sentiment']]


merged_df = pd.concat([amazon_df, disaster_df, judge_df, senti_df], ignore_index=True)


merged_df.to_csv("merged_dataset.csv", index=False)
```
